* 이 코드는 텐서플로 최신 버전에 맞춰 갱신되어 있습니다.
* 1·2쇄에 인쇄된 코드와 일부 다른 곳이 있습니다.(3쇄부터는 책과 같습니다.)
* 자료와 정오표 → mlfundamentals.github.io/ai-structure

```python
import tensorflow as tf

w = tf.Variable(tf.random.normal([1]), name='w')
b = tf.Variable(tf.random.normal([1]), name='b')

X = tf.constant([1, 2, 3], dtype=tf.float32)
Y = tf.constant([1, 2, 3], dtype=tf.float32)

hypothesis = lambda: w * X + b

cost = lambda: tf.reduce_mean(tf.square(hypothesis() - Y))

optimizer = tf.optimizers.SGD(learning_rate=0.01)

print("학습횟수    weight      bias      cost")
for step in range(2001):
    with tf.GradientTape() as tape: loss = cost()
    optimizer.apply_gradients(zip(tape.gradient(loss, [w, b]), [w, b]))
    if step % 100 == 0:
        print(format(step, '4d'), "  ", format(w.numpy()[0], '10.4f'), " ", format(b.numpy()[0], '8.4f'), format(cost().numpy(), '10.5f'))

print("\n사과를 5개 사면? ", format(w.numpy()[0] * 5 + b.numpy()[0], '5.3f'), "원")
```
